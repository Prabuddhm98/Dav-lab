```python
grade_info = pd.DataFrame({
    'GradeClass':[0,1,2,3,4],
    'Grade':['A','B','C','D','F'],
    "grade_description":['Excelient','Good','Average','Below average','Fail']
})

print(grade_info)
```

```python
merged = pd.merge(
    df,
    grade_info,
    on = 'GradeClass',
    how = 'left'
)
print("\n Merged Dataframe:")
print(merged.head())
```

```python
grade_details = pd.DataFrame({
    'description':[
        'Excelient',
        'Good',
        "Average",
        'Below Average',
        'Fail'
    ],
    'performance_level':[
        'very High',
        'High',
        'Medium',
        'Low',
        'Very Low'
    ]
},index = [0, 1, 2, 3, 4])
print(grade_details.head())
```

```python
grade_counts = df['GradeClass'].value_counts().sort_index().to_frame('student_count')
print(grade_counts.head())

joined = grade_counts.join(grade_details)
print(joined)
```

```python
df1 = df.iloc[:100]
df2 = df.iloc[100:200]
combined = pd.concat([df1,df2])
print(combined.head())
print(combined.shape)
```

```python
average_score = df.groupby('GradeClass')['GPA'].mean()
print("\n Average GPA by Grade:")
print(average_score)
```

```python
pivot = pd.pivot_table(
    df,
    values='GPA',
    index = 'GradeClass',
    aggfunc='mean'
)
print("\nPivot Table - Average GPA by Grade:")
print(pivot)
```
