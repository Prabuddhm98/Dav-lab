```python
df_mean = df.copy()

mean_building_area = df_mean['BuildingArea'].mean()

df_mean["BuildingArea"] = df_mean["BuildingArea"].fillna(
    mean_building_area
)

print("\n missing building area values after mean Imoutation ")
print(df_mean["BuildingArea"].isnull().sum())
```

```python
df_median = df.copy()

median_year_built = df_median["YearBuilt"].median()

df_median["YearBuilt"]  =df_median["YearBuilt"].fillna(
    median_year_built
)

print("\n Median Year Built",median_year_built)

print("\n Missing YearBuilt values after median imputation")
print(df_median["YearBuilt"].isnull().sum())
```

```python
df_ffill = df.copy()
df_ffill["Car"] = df_ffill["Car"].ffill()

print("\n Missing car value after forward fill")

print(df_ffill["Car"].isnull().sum())
```

```python
comparison = pd.DataFrame({
    " Original": df["BuildingArea"].head(15).values,
    "Mean Imputed": df_mean["BuildingArea"].head(15).values,

    "Original": df["YearBuilt"].head(15).values,
    "Median Imputed": df_median["YearBuilt"].head(15).values,

    "Original": df["Car"].head(15).values,
    "Forward Filled": df_ffill["Car"].head(15).values
})


print("\nComparison of Imputation Methods")
print(comparison)
```

```python
df_processed = df.copy()

council_mode = df_processed["CouncilArea"].mode()[0]

df_processed["CouncilArea"] = df_processed["CouncilArea"].fillna(
    council_mode
)

print("\nCouncil area mode:", council_mode)

print("\nMissing categorical values:")
print(df_processed[["CouncilArea"]].isnull().sum())
```

```python
df_encoded = pd.get_dummies(
    df_processed,
    columns=["Type", "Method"],
    dtype=int
)

print("\nDocument after one-hot encoding:")
print(df_encoded.head())
```

```python

df_label = df.copy()

label_encoder = LabelEncoder()

df_label["CouncilArea_Encoded"] = label_encoder.fit_transform(
    df_label["CouncilArea"]
)

print("\nLabel encoding of council area")

print(
    df_label[
        ["CouncilArea", "CouncilArea_Encoded"]
    ].head(10)
)
```
