```python
import numpy as np
arr = np.array([ 10,20,30,40,50])
print("first element:",arr[0])
print("third element:",arr[2])
print("last element:",arr[-1])
```

```python
print("----basic array operation----")
```

```python
print("add 5:",arr + 5)
print("subtract 5:",arr - 5)
print("multiply by 2:",arr * 2)
print("divide by 10:",arr/10)
```

```python
print("----mathematical operation----")
```

```python
print("sum:",np.sum(arr))
print("mean:",np.mean(arr))
print("maximum:",np.max(arr))
print("minimum:",np.min(arr))
```

```python
import pandas as pd
```

```python
data = {
    "name":["prasad","ravi","mahtab","mahesh"],
    "marks":[88,72,91,65],
    "branch":["CSE","ECE","CSE","CSE"]
}
```

```python
df=pd.DataFrame(data)

print("Pandas data frame")
print(df)
```

```python
print("----Column Indexing----")
```

```python
print("Name Column:")
print(df["name"])


print("Marks column:")
print(df["marks"])
```

```python
print("---row indexng----")
```

```python
print("First Row:")
print(df.iloc[0])

print("Third Row:")
print(df.iloc[2])
```

```python
print("---specific value---")

print("Marks of first student is:",df.iloc[0,1])
```

```python
print("---row slicing---")

print("first tow rows:")
print(df.iloc[0:2])
```

```python
print("---row and column sclicing----")

print("First three rows with name and marks:")
print(df.iloc[0:3,[0,1]])
```

```python
print("----filtering data----")

result = df[df["marks"] > 70]

print("student with marks greater than 70:")
print(result)
```
